```python

# from pathlib import Path
# import pandas as pd
#
# root = Path("/home/rpvselva/IdeaProjects/papiIla/PapilaDB-PAPILA-9c67b80983805f0f886b068af800ef2b507e7dc0")
# arquivo = root / "ClinicalData" / "patient_data_od.xlsx"
#
# print("Arquivo encontrado:", arquivo.is_file())
# df = pd.read_excel(arquivo)
# display(df.head())
# from pathlib import Path
# import sys
#
# root = Path("/home/rpvselva/IdeaProjects/papiIla/PapilaDB-PAPILA-9c67b80983805f0f886b068af800ef2b507e7dc0")
# sys.path.insert(0, str(root))
#
# from papila_dataset import load_clinical_data
#
# df_od, df_os = load_clinical_data(root)
#
# print("OD (olho direito):", df_od.shape)
# print("OS (olho esquerdo):", df_os.shape)
#
# display(df_od.head())
# display(df_os.head())
"""
Módulo auxiliar para carregar os dados clínicos do PAPILA.

Uso:
    from papila_dataset import load_clinical_data
    df_od, df_os = load_clinical_data(root_dir)
"""
import re
from pathlib import Path

import pandas as pd


# Nomes das colunas após pular os 2 cabeçalhos
COLUNAS = [
    "Patient_ID_Raw", "Age", "Gender", "Diagnosis",
    "dioptre_1", "dioptre_2", "astigmatism", "Phakic/Pseudophakic",
    "Pneumatic", "Perkins", "Pachymetry", "Axial_Length", "VF_MD",
]


def _carregar_um_olho(xlsx_path: Path) -> pd.DataFrame:
    """Carrega metadados de um olho (OD ou OS)."""
    df = pd.read_excel(xlsx_path, header=None)
    df = df.iloc[2:].reset_index(drop=True)
    df.columns = COLUNAS

    # Extrair Patient_ID numérico (#002 → 002)
    df["Patient_ID"] = df["Patient_ID_Raw"].apply(
        lambda v: re.match(r"^#(\d+)$", str(v).strip()).group(1).zfill(3)
        if pd.notna(v) and re.match(r"^#(\d+)$", str(v).strip())
        else None
    )
    df = df.dropna(subset=["Patient_ID"]).reset_index(drop=True)

    # Diagnóstico numérico → texto
    mapa = {0: "Normal (NON)", 1: "Glaucoma Suspect (SUSP)",
            2: "Glaucomatous Optic Neuropathy (GON)"}
    df["Diagnosis_Label"] = df["Diagnosis"].map(mapa)

    return df


def load_clinical_data(root: Path) -> tuple:
    """
    Carrega os dados clínicos do PAPILA.

    Args:
        root: diretório raiz do PAPILA (contém ClinicalData/)

    Returns:
        (df_od, df_os): DataFrames dos olhos direito e esquerdo
    """
    root = Path(root)

    # Procurar os Excels
    od_paths = list(root.rglob("patient_data_od.xlsx"))
    os_paths = list(root.rglob("patient_data_os.xlsx"))

    if not od_paths or not os_paths:
        raise FileNotFoundError(
            f"Excels não encontrados em {root}. "
            f"Verifique se o PAPILA foi baixado corretamente."
        )

    df_od = _carregar_um_olho(od_paths[0])
    df_os = _carregar_um_olho(os_paths[0])

    return df_od, df_os


if __name__ == "__main__":
    #root = Path(__file__).resolve().parent
    root = Path("/home/rpvselva/IdeaProjects/papiIla/PapilaDB-PAPILA-9c67b80983805f0f886b068af800ef2b507e7dc0")
    df_od, df_os = load_clinical_data(root)
    print("OD:", df_od.shape)
    print("OS:", df_os.shape)
    print(df_od.head())
    print(df_os.head())
```
